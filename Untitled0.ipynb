Mutability in Python refers to whether an object can be modified after it is created:
*   Mutable objects: Can be changed. Examples: list, dict, set.
*   • Immutable objects: Cannot be changed. Examples: int, float, str, tuple.
For instance, modifying a list will change its content, while modifying a string creates a new

#6
In Python, lists are mutable and tuples are immutable due to their design and purpose:
1. Lists (mutable): Designed for dynamic operations like adding, removing, or
changing elements. This makes them flexible for use cases requiring frequent updates.
my_list = [1, 2, 3]
my_list[0] = 10 # Allowed
print(my_list) # Output: [10, 2, 3]
2. Tuples (immutable): Optimized for performance and designed to store fixed,
unchangeable data. This makes them faster and suitable for use as keys in dictionaries or in
scenarios where data integrity is important.
my_tuple = (1, 2, 3)
my_tuple[0] = 10 # Error: 'tuple' object does not support item assignment
This distinction is intentional to balance flexibility and performance in different use cases.

#7
The difference between == and is in Python is:
1. == (Equality Operator):
• Checks if the values of two objects are equal.
• Does not consider whether they are the same object in memory.
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b) # True (values are the same)
print(a is b) # False (different objects in memory)
2. is (Identity Operator):Checks if two objects refer to the same memory location (i.e., they are the same
object).
a = [1, 2, 3]
c = a
print(a is c) # True (same object )

#9
Type casting in Python converts one data type to another. It can be:

Implicit: Python automatically converts types when no data loss occurs, like converting an int to a float.
Example: result = 5 + 3.2 (int becomes float).

Explicit: Manually convert types using functions like int(), float(), or str().
Example: y = int("10") converts a string to an integer.

#10
1. Implicit Type Casting:
   - Python automatically converts a smaller data type to a larger data type.
   - Happens when there is no loss of information.

   Example:
   python
   x = 5       # int
   y = 3.2     # float
   result = x + y  # int is implicitly converted to float
   print(result)  # 8.2 (float)
   

2. Explicit Type Casting:
   - You manually convert one data type to another using functions like int(), float(), or str().
   - Requires explicit code.

   Example:
   python
   x = "10"      # string
   y = int(x)    # manually convert string to int
   print(y + 5)  # 15 (int)

The purpose of conditional statements in Python is to allow decision-making in code. They enable the execution of different blocks of code based on specific conditions, making the program more dynamic.

- if: Executes a block of code if the condition is true.
- elif: Tests additional conditions if the previous if (or elif) was false.
- else: Executes a block of code if none of the conditions are true.

Example:
python
x = 10
if x > 5:
    print("Greater than 5")
else:
    print("5 or less")

This allows the program to respond to different situations based on input or other factors.

#12
The elif statement checks multiple conditions after an if statement. If the if condition is false, it checks the elif condition. If elif is true, its block executes; otherwise, it moves to the next elif or else.

Example:
python
x = 10
if x > 15:
    print("Greater than 15")
elif x > 5:
    print("Greater than 5")
else:
    print("5 or less")
Output: "Greater than 5"

#13
1. for loop:
   - Used when the number of iterations is known or when iterating over a sequence (like a list, tuple, or range).
   - Syntax: for item in sequence:
   Example:
   python
   for i in range(5):
       print(i)
   

2. while loop:
   - Used when the number of iterations is not known and depends on a condition being true.
   - Syntax: while condition:
   Example:
   python
   i = 0
   while i < 5:
       print(i)
       i += 1
The for loop is ideal for iterating over a fixed range or sequence, while the while loop is used when the iteration depends on a condition.

#14
A while loop is more suitable when the number of iterations is not known in advance and depends on a dynamic condition.
Example Scenario:
Waiting for a user to input the correct password.

python
password = ""
while password != "correct":
    password = input("Enter the password: ")
print("Access granted!")
Here, the loop continues until the correct password is entered, and the number of iterations is not fixed.

```python
#6
def test_mutability(lst, val):
    lst.append(4)  # Mutable object
    val = 10  # Immutable object
    print(f"List after modification: {lst}")
    print(f"Value after modification: {val}")

lst = [1, 2, 3]
val = 5
test_mutability(lst, val)
print(f"List outside function: {lst}")
print(f"Value outside function: {val}")
```

```python
#7
def modify_elements(lst, val):
    lst.append(10)  # Mutable object
    val = 100  # Immutable object
    print(f"Modified list: {lst}")
    print(f"Modified value: {val}")

my_list = [1, 2, 3]
my_value = 50
modify_elements(my_list, my_value)
print(f"List after function: {my_list}")
print(f"Value after function: {my_value}")
```

```python
#14
my_string = input("Enter a string: ")
reversed_string = ""
index = len(my_string) - 1
while index >= 0:
    reversed_string += my_string[index]
    index -= 1
print(f"Reversed string: {reversed_string}")
```
